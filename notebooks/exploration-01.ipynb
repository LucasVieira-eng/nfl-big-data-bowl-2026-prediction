```python
def string_cols(df, skip=("source_file",)):
    return [c for c in df.columns
            if pd.api.types.is_string_dtype(df[c]) and c not in skip]

str_cols = string_cols(train_input)

# 1) number of unique values per string column
print(train_input[str_cols].nunique().sort_values())
```
